# Problem: Implement Linear Regression

### Problem Statement
Your task is to implement a **Linear Regression** model using PyTorch. The model should predict a continuous target variable based on a given set of input features.

### Requirements
1. **Model Definition**:
   - Implement a class `LinearRegressionModel` with:
     - A single linear layer mapping input features to the target variable.
2. **Forward Method**:
   - Implement the `forward` method to compute predictions given input data.

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
# Generate synthetic data
torch.manual_seed(42) #makes the random vlaues reproducible 
X = torch.rand(100, 1) * 10  # 100 data points between 0 and 10
y = 2 * X + 3 + torch.randn(100, 1)  # Linear relationship with noise

# Define the Linear Regression Model
#TODO: Add the layer and forward implementation
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1,1) #creates a layer with one input and one output

    def forward(self, x):
        return self.linear(x) #sends the input through that layer and returns its predictions

# Initialize the model, loss function, and optimizer
model = LinearRegressionModel()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01) 
#SGD = stochastic gradient descent : adjusts the model's parameters in the direction that should reduce the error. (uses LR)

# Training loop
epochs = 1000
for epoch in range(epochs):
    # Forward pass
    predictions = model(X) #draw the current line's predictions
    loss = criterion(predictions, y) #check how far the line is from teh data

    # Backward pass and optimization
    optimizer.zero_grad() #clear the gradients from previous attempt
    loss.backward() #calculates the gradients: figures out how the weight and bias should change to reduce loss
    optimizer.step() #uses those gradients

    # Log progress every 100 epochs
    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch + 1}/{epochs}], Loss: {loss.item():.4f}")
```

## Interview Questions

1. The loss stops near 0.66 instead of reaching 0. Why can't a perfeect implementation get it to zero, and what in the data-generation line decides where it levels off?
    
    The Y values contain random noise. The model would be able to learn the straight line pattern but not the noise. Therefore, the loss won't ever reach 0
2. What goes wrong if you delete optimizer.zero_grad()?
    Without it, each call to loss.backwards() adds teh new gradients to the ones already stores. Then optimizer.step() applies the accumulated gradients. The updates will become to large and teh loss will explode up.
3. Linear Regression has an exact closed-from answer. Why woudl someone use gradient descent for it instead?
    Gradient descent works best for much larger or more complex models where a direct formula may be impractical. For a smaller linear regression example, the direct forumula approach is simpler.


```python
# Display the learned parameters
[w, b] = model.linear.parameters()
print(f"Learned weight: {w.item():.4f}, Learned bias: {b.item():.4f}")

# Testing on new data
X_test = torch.tensor([[4.0], [7.0]])
with torch.no_grad():
    predictions = model(X_test)
    print(f"Predictions for {X_test.tolist()}: {predictions.tolist()}")
```
